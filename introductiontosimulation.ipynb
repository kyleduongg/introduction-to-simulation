To start, the first function that will be explored is: **`np.random.choice()`**!

### `np.random.choice()`

The `np.random.choice()` function allows us to randomly select an outcome from a given collection of possible values. This is particularly useful when simulating random experiments, such as flipping coins, rolling dice, or drawing cards.

Before we begin using this function, let's take a look at its syntax:

At first glance, this might seem a little confusing. However, each parameter serves a specific purpose.

| Parameter | Description | Default Value |

| `p` | Specifies the probability of selecting each possible outcome. | `None` (equal probabilities) |

### **Exercise 2.3: Sampling With and Without Replacement**

As of right now, we've learned how to use `np.random.choice()` to generate a single random outcome and multiple random outcomes using the `size` parameter.

Now, what if we **don't want the same outcome to be selected mroe than once?**

This is where the `replace` parameter comes in! Recall that `np.random.choice()` has a parameter called `replace`, which determines whether an item can be selected more than once.

There are two ways we can sample:

- **With Replacement (`replace=True`):** After selecting an item, we put it back before making another selection. This means the same item can be selected multiple times.
- **Without Replacement (`replace=False`):** After selecting an item, we do not put it back. This means the same item cannot be selected again.

By design, NumPy sets `np.random.choice()` behavior to replace=True, meaning that sampling is performed with replacement unless specified otherwise.

#### **Let's Look at an Example!**

Imagine a bag containing four M & M's, each with a different color:


- <span style="color: #FF5252;">Red</span>
- <span style="color: #42A5F5;">Blue</span>
- <span style="color: #66BB6A;">Green</span>
- <span style="color: #FBC02D;">Yellow</span>


Suppose we want to randomly draw **3 M & M's** from this bag. There are two ways we can draw M&Ms from the bag:

- **With Replacement (`replace=True`):** After drawing an M&M, we **put it back** into the bag before the next draw. This means the same M&M can be selected multiple times.

- **Without Replacement (`replace=False`):** After drawing an M&M, we **keep it out** of the bag. This means that the same M&M cannot be selected again.

### **2.5. Supplementary Notes**

Now that we've learned how to generate random outcomes using `np.random.choice()`, let's explore a few useful functions for **analyzing simulation results**!

| Function | Description |
|:---|:---|
| `np.sum()` | Calculates the sum of values or counts outcomes that satisfy a condition. |
| `np.mean()` | Calculates the average or proportion of outcomes satisfying a condition. |
| `len()` | Returns the number of elements in a collection. |
| `np.unique()` | Identifies unique values and can count their occurrences using `return_counts=True`. |

### Let's Look at an Example!

Suppose we simulated rolling a fair six-sided die **10 times**. These functions are useful to calculate the total, count specific outcomes, find the average, and analyze how frequently each number appears.

````python
# Simulate rolling a fair six-sided die 10 times & print their results
rolls = np.random.choice([1, 2, 3, 4, 5, 6], size=10)
print("Simulated rolls:", rolls)

# Sum of all die rolls
print("Sum:", np.sum(rolls))

# Number of times we rolled a six
print("Number of sixes:", np.sum(rolls == 6))

# Average die roll
print("Average:", np.mean(rolls))

# Total number of rolls
print("Total rolls:", len(rolls))

# Proportion of rolls that were sixes
print("Proportion of sixes:", np.mean(rolls == 6))

# Unique outcomes and their counts
values, counts = np.unique(rolls, return_counts=True)

print("Unique values:", values)
print("Counts:", counts)
````

## **3. Introduction to `np.random.randint()`**


Let's explore another useful NumPy function: **`np.random.randint()`**.

### `np.random.randint()`

The `np.random.randint()` function generates **random integers (whole numbers)** within a specified range. Unlike `np.random.choice()`, it generates random numbers **without listing every possible value**.

### Understanding the Syntax

Let's take a look at the main parameters:

`np.random.randint(low, high=None, size=None)`

| Parameter | Description | Default Value |
|:---|:---|:---|
| `low` | The smallest possible integer (**included**) when `high` is provided. | Required |
| `high` | The upper limit (**not included**). | `None` |
| `size` | The number of random integers to generate. | `None` (one integer) |

**Important Things to Remember:**

- **`low` is inclusive:** The smallest possible number can be generated.
- **`high` is exclusive:** The upper limit will never be generated.
- **If only one number is provided:** NumPy starts at 0 and stops before that number.

For example, `np.random.randint(1, 7)` generates integers from **1 to 6**, while `np.random.randint(6)` generates integers from **0 to 5**.

Let's see how this works in a simulation now!

## **4. Introduction to `np.random.rand()`**

Now, let's explore our another NumPy random function: **`np.random.rand()`**.

### `np.random.rand()`

The `np.random.rand()` function generates **random decimal numbers between 0 and 1**. Unlike `np.random.randint()`, which generates whole numbers, `np.random.rand()` generates decimal values.

### Understanding the Syntax

Let's take a look at how this function works:

`np.random.rand(d0, d1, ..., dn)`

| Parameter | Description |
|:---|:---|
| `d0` | The number of random values to generate in one dimension. |
| `d1, ..., dn` | Optional additional dimensions for organizing the random values into an array. |

**Important Things to Remember:**

- **0 is included**, but **1 is excluded**.
- Every value between 0 and 1 is equally likely to be generated, assuming a uniform distribution.
- By default, `np.random.rand()` generates a single random decimal number.

For example, `np.random.rand(5)` generates **5 random decimal numbers between 0 and 1**.

Now, let's see how we can use these numbers to simulate random events!

````python
# Simulate the weather over 100 days
random_numbers = np.random.rand(100)

# Determine which days are rain
rainy_days = random_numbers < 0.3

# Calculate the proportion of rainy days
estimated_probability = np.mean(rainy_days)

# DO NOT MODIFY ANYTHING BELOW! 

results_table = pd.DataFrame({
    "Statistic": [
        "Number of rainy days",
        "Proportion of rainy days",
        "Theoretical probability"
    ],
    "Value": [
        np.sum(rainy_days),
        round(estimated_probability, 3),
        0.3
    ]
})

display(results_table)
````

## **5. Building Simulations with Python**

Now that we know how to generate random outcomes using NumPy, let's explore how to **build more complex simulations**!

Creating simulations often involves **repeating experiments, tracking results, and stopping when certain conditions are met**. Python provides several useful tools to help us accomplish this:

| Python Tool | Purpose |
|:---|:---|
| `if` / `else` | Make decisions based on a condition. |
| `for` loops | Repeat an experiment a specified number of times. |
| `while` loops | Continue repeating an experiment while a condition is true. |
| `break` | Stop a loop when a particular condition is met. |
| `.append()` | Store outcomes in a list as a simulation runs. |

Let's see how we can use these tools to build our own simulations!



### **5.1. Making Decisions with `if` and `else`**

When running a simulation, we often want to **perform different actions based on the outcome**. Python allows us to do this using **conditional statements**.

| Statement | Purpose |
|:---|:---|
| `if` | Executes code when a condition is **True**. |
| `else` | Executes code when the condition is **False**. |

### Let's Look at an Example!

Imagine rolling a fair six-sided die where **rolling a 6 wins a prize**. Any other number means we don't win.

### **5.2. Repeating Experiments with `for` Loops**

Let's imagine flipping a coin **100 times**. Writing the same line of code 100 times wouldn't be very efficient!

Instead, Python provides **`for` loops**, which allow us to repeat an experiment a specified number of times.

For example, `for i in range(5):` repeats the indented code **5 times**.

We can also use two useful Python tools to keep track of our results:

| Python Tool | Purpose |
|:---|:---|
| `[]` | Creates an empty list to store our outcomes. |
| `.append()` | Adds each new outcome to the list. |

By combining a `for` loop with a list, we can **repeat a simulation and record its results**. 

Let's see how this works!


````python
# Create an empty list to store the outcomes
flips = []

# Simulate flipping a coin 5 times
for i in range(5):
    flip = np.random.choice(["Heads", "Tails"])
    flips.append(flip)

# Display the results
print("Coin flips:", flips)
print("Total flips:", len(flips))
````

You should have noticed that the simulation generated **5 coin flips** and stored the outcomes in a list.

Here's what each part of the code does:

- **`flips = []`** creates an empty list to store our results.
- **`for i in range(5):`** repeats the experiment 5 times.
- **`flips.append(flip)`** adds each new outcome to the list.

This allows us to **repeat and record random experiments** without writing the same code multiple times.

### **Exercise 5.1: Counting Heads with a Loop**

Let's take our coin-flipping simulation one step further!

Suppose we want to simulate **20 coin flips** and determine how many times the coin lands on heads.

<span style="display: inline-block; background-color: #eff6ff; color: #1e40af; border-left: 4px solid #3b82f6; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">✏️ Your Turn!</span>

**Question 5.1:** Using a `for` loop and `np.random.choice()`, simulate flipping a fair coin **20 times** and count how many times heads appears.

**Instructions:**
- Use a `for` loop to generate 20 coin flips and store the outcomes in a list called `coin_flips`.
- Use an `if` statement to count the number of heads, storing the result in `heads_count`.

<span style="display: inline-block; background-color: #fffbeb; color: #92400e; padding: 8px 12px; border-left: 4px solid #f59e0b; border-radius: 5px;">

### **5.3. Repeating Until a Condition Is Met**

Currently, we learned that `for` loops allow us to repeat an experiment a **specified number of times**. However, some experiments require us to keep repeating until a particular outcome occurs.

This is where **`while` loops** come in!

| Python Tool | Purpose |
|:---|:---|
| `while` | Repeats code as long as a condition remains true. |
| `while True` | Continuously repeats code until the loop is stopped. |
| `break` | Immediately stops the loop when a condition is met. |

---


### Let's Look at an Example!

Imagine rolling a fair six-sided die repeatedly **until we roll a 6**.

We don't know how many rolls it will take, so we'll use:

- **`while`** to keep rolling the die.
- **`break`** to stop the simulation once we roll a 6.

Let's see how many rolls it takes!


````python
# Initialize the number of rolls
roll_count = 0

# Keep rolling until we get a 6
while True:
    roll = np.random.randint(1, 7)
    roll_count += 1

    if roll == 6:
        break

# Display the results
print("Final roll:", roll)
print("Number of rolls needed:", roll_count)
````

---


<span style="display: inline-block; background-color: #ecfdf5; color: #065f46; border-left: 4px solid #10b981; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">🔍 What Did We Observe?</span>

You should have noticed that the final roll was always **6**, but the number of attempts could vary each time!

Here's how the code works:

- **`while True:`** keeps repeating the simulation.
- **`roll_count += 1`** increases the counter by 1 after each roll.
- **`if roll == 6:`** checks whether we rolled a 6.
- **`break`** stops the loop once the condition is met.


Remember that `while True` will continue running indefinitely unless a stopping condition, such as `break`, is included!

### **Exercise 5.2: Flipping Until Heads**

Imagine repeatedly flipping a fair coin **until heads appears for the first time**.

Unlike our previous simulations, we don't know how many flips it will take! Let's use what we've learned about `while` loops to find out.

<span style="display: inline-block; background-color: #eff6ff; color: #1e40af; border-left: 4px solid #3b82f6; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">✏️ Your Turn!</span>

**Question 5.2:** Using a `while` loop and `np.random.choice()`, simulate flipping a fair coin **until heads appears for the first time**.

**Instructions:**
- Repeatedly flip a coin using a `while` loop, storing each outcome in a list called `flips`.
- Stop the simulation when heads appears, then use `len()` to determine the total number of flips.

<span style="display: inline-block; background-color: #fffbeb; color: #92400e; padding: 8px 12px; border-left: 4px solid #f59e0b; border-radius: 5px;">
